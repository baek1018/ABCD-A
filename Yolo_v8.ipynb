```python
!pip install ultralytics
from ultralytics import YOLO
from google.colab import files

# YOLOv8 모델 불러오기 (고양이도 인식 가능한 모델)
model = YOLO('yolov8n.pt')  # 모델을 "yolov8s.pt" 또는 "yolov8m.pt" 등으로 변경 가능

# 이미지 업로드
uploaded = files.upload()

# 업로드한 이미지 파일의 이름을 확인하고, 경로 설정
image_path = list(uploaded.keys())[0]

# 이미지에 대해 객체 탐지 실행
results = model(image_path)

# 결과 출력
results[0].show() # Access the first result object in the list and call show

# 고양이 클래스의 ID는 일반적으로 COCO 데이터셋에서 'cat'이므로 ID=16입니다.
# 탐지된 고양이 수 출력
# Convert the results to a Pandas DataFrame using .to_df()
results_df = results[0].to_df()

# Filter the DataFrame to count detections with class ID 16 (cat)
cats_detected = len(results_df[results_df['class'] == 16])

print(f"탐지된 고양이 수: {cats_detected}")
```
